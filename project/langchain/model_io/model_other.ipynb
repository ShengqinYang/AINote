```python
from langchain_community.llms.xinference import Xinference
from langchain.prompts import PromptTemplate
from langchain_community.embeddings import XinferenceEmbeddings

llm_model_uid = "local2_qwen1.5" # 通过xinference启动的llm模型 model_id
emb_model_uid = "text2_local" # 通过xinference启动的embedding模型 model_id
```

```python
llm = Xinference(server_url = "http://localhost:9997", model_uid = llm_model_uid)
embed = XinferenceEmbeddings(
    server_url = "http://127.0.0.1:9997", 
    model_uid = emb_model_uid # model_uid is the uid returned from launching the model
)
```

```python

def use_lcoal_llm():
    template = """Q: 小说《{novel_name}》的作者是谁？只说名字

    A: """

    prompt_template = PromptTemplate.from_template(template)
    print(prompt_template.input_variables)
    # 格式化 PromptTemplate
    formatted_prompt = prompt_template.format(novel_name = "老人与海")
    print(formatted_prompt,"-----------")
    response = llm(prompt = formatted_prompt, stop = ["Q:", "\n"])
    return response
use_lcoal_llm()
```

```python
def use_local_embed():
    text = ["Q：《老人与海》的作者是谁？A：海名威"]
    q_str = "海名威有什么著作？"
    # embed_text = embed.embed_documents(text)
    # print(embed_text)

    query_result = embed.embed_query(q_str)
    print(query_result,"query_result")
    return query_result
use_local_embed()
```
